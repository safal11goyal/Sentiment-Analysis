```python
import pandas as pd
import re
import string
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download NLTK resources (only need to run once)
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')

# ============================
# 1. Load Dataset
# ============================
train_path = "../data/processed/train.csv"
test_path = "../data/processed/test.csv"

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
print(train_df.head())

# ============================
# 2. Text Cleaning Functions
# ============================
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

def clean_text(text, method="lemmatize"):
    # Lowercase
    text = text.lower()
    # Remove HTML tags
    text = re.sub(r'<.*?>', ' ', text)
    # Remove URLs
    text = re.sub(r'http\S+|www\S+', ' ', text)
    # Remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # Tokenize
    tokens = text.split()
    # Remove stopwords
    tokens = [word for word in tokens if word not in stop_words]
    # Stemming or Lemmatization
    if method == "stem":
        tokens = [stemmer.stem(word) for word in tokens]
    else:
        tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return " ".join(tokens)

# ============================
# 3. Apply Preprocessing
# ============================
train_df['clean_review'] = train_df['review'].apply(lambda x: clean_text(str(x)))
test_df['clean_review'] = test_df['review'].apply(lambda x: clean_text(str(x)))

print("\nSample cleaned text:")
print(train_df[['review','clean_review']].head())

# ============================
# 4. Save Cleaned Data
# ============================
train_df.to_csv("../data/processed/train_clean.csv", index=False)
test_df.to_csv("../data/processed/test_clean.csv", index=False)

print("✅ Preprocessing completed! Cleaned files saved in ../data/processed/")
```
