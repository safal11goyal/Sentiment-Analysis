```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import seaborn as sns
import matplotlib.pyplot as plt
import pickle

# ============================
# 1. Load Preprocessed Data
# ============================
train_path = "../data/processed/train_clean.csv"
test_path = "../data/processed/test_clean.csv"

train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)

X_train = train_df['clean_review']
y_train = train_df['sentiment']
X_test = test_df['clean_review']
y_test = test_df['sentiment']

print("Train size:", X_train.shape)
print("Test size:", X_test.shape)

# ============================
# 2. Vectorization (TF-IDF)
# ============================
tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1,2))

X_train_vec = tfidf.fit_transform(X_train)
X_test_vec = tfidf.transform(X_test)

print("TF-IDF shape:", X_train_vec.shape)

# ============================
# 3. Train Model (Naive Bayes)
# ============================
model = MultinomialNB()
model.fit(X_train_vec, y_train)

# ============================
# 4. Evaluation
# ============================
y_pred = model.predict(X_test_vec)

print("\nAccuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6,4))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=model.classes_, yticklabels=model.classes_)
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Confusion Matrix")
plt.show()

# ============================
# 5. Save Model & Vectorizer
# ============================
with open("../models/naive_bayes.pkl", "wb") as f:
    pickle.dump(model, f)

with open("../models/tfidf_vectorizer.pkl", "wb") as f:
    pickle.dump(tfidf, f)

print("✅ Model and TF-IDF vectorizer saved in ../models/")
```
