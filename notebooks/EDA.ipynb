```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud, STOPWORDS

# ============================
# 1. Load Dataset
# ============================
# Change the path to your dataset (e.g., IMDB reviews CSV)
data_path = "../data/raw/train.csv"  # adjust path as needed
df = pd.read_csv(data_path)

# Preview data
print("Dataset shape:", df.shape)
print(df.head())

# ============================
# 2. Basic Info
# ============================
print("\nData Info:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nClass Distribution:")
print(df['sentiment'].value_counts())

# Plot class distribution
plt.figure(figsize=(6,4))
sns.countplot(x='sentiment', data=df, palette='Set2')
plt.title('Sentiment Distribution')
plt.show()

# ============================
# 3. Text Length Analysis
# ============================
df['text_length'] = df['review'].apply(len)

plt.figure(figsize=(8,4))
sns.histplot(df[df['sentiment']=="positive"]['text_length'], color="green", label="Positive", kde=True)
sns.histplot(df[df['sentiment']=="negative"]['text_length'], color="red", label="Negative", kde=True)
plt.legend()
plt.title("Text Length Distribution")
plt.show()

# ============================
# 4. WordClouds
# ============================
stopwords = set(STOPWORDS)

# Positive reviews wordcloud
pos_text = " ".join(df[df['sentiment']=="positive"]['review'])
wordcloud_pos = WordCloud(width=800, height=400, stopwords=stopwords, background_color='white').generate(pos_text)

plt.figure(figsize=(10,5))
plt.imshow(wordcloud_pos, interpolation='bilinear')
plt.axis('off')
plt.title("WordCloud - Positive Reviews")
plt.show()

# Negative reviews wordcloud
neg_text = " ".join(df[df['sentiment']=="negative"]['review'])
wordcloud_neg = WordCloud(width=800, height=400, stopwords=stopwords, background_color='black', colormap='Reds').generate(neg_text)

plt.figure(figsize=(10,5))
plt.imshow(wordcloud_neg, interpolation='bilinear')
plt.axis('off')
plt.title("WordCloud - Negative Reviews")
plt.show()

# ============================
# 5. Most Frequent Words
# ============================
from collections import Counter
import re

# Simple tokenizer
def tokenize(text):
    text = text.lower()
    tokens = re.findall(r"\b[a-z]{3,}\b", text)  # only words with 3+ letters
    return tokens

# Positive words frequency
pos_words = Counter([word for review in df[df['sentiment']=="positive"]['review'] for word in tokenize(review)])
neg_words = Counter([word for review in df[df['sentiment']=="negative"]['review'] for word in tokenize(review)])

# Top 20 words
pos_common = pd.DataFrame(pos_words.most_common(20), columns=['word','count'])
neg_common = pd.DataFrame(neg_words.most_common(20), columns=['word','count'])

fig, axes = plt.subplots(1,2, figsize=(14,6))
sns.barplot(y='word', x='count', data=pos_common, ax=axes[0], palette='Greens_r')
axes[0].set_title("Top Positive Words")

sns.barplot(y='word', x='count', data=neg_common, ax=axes[1], palette='Reds_r')
axes[1].set_title("Top Negative Words")

plt.tight_layout()
plt.show()
```
